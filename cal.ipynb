```python
import os
import pandas as pd

# 定义存放CSV文件的目录路径
directory_path = r"D:\Python code\oggm\mb-cal"

# 处理每个CSV文件的函数
def process_csv(file_path):
    # 读取CSV文件
    df = pd.read_csv(file_path)
    
    # 删除已经存在的 'Multi_Year_Avg_Specific_MB' 列（如果存在）
    if 'Multi_Year_Avg_Specific_MB' in df.columns:
        df = df.drop(columns=['Multi_Year_Avg_Specific_MB'])
        print(f"已删除 {file_path} 中的 'Multi_Year_Avg_Specific_MB' 列。")
    
    # 检查是否包含所需的列
    if 'Year' in df.columns and 'Specific_MB' in df.columns:
        # 筛选出2000到2020年之间的数据
        df_filtered = df[(df['Year'] >= 2000) & (df['Year'] <= 2020)]
        
        # 计算2000-2020年期间的 'Specific_MB' 多年平均值
        mean_specific_mb = df_filtered['Specific_MB'].mean()
        
        # 只在第一行添加这一列的值
        df.at[0, 'Multi_Year_Avg_Specific_MB'] = mean_specific_mb
        
        # 保存修改后的数据框回到CSV文件
        df.to_csv(file_path, index=False)
        print(f"已处理 {file_path}，在第一行添加了 'Multi_Year_Avg_Specific_MB' 列。")
    else:
        print(f"跳过 {file_path}，因为文件中缺少所需的列。")

# 遍历目录中的所有文件
for filename in os.listdir(directory_path):
    if filename.endswith(".csv") and filename.startswith("mb_ts_RGI60-13"):
        # 构造文件的完整路径
        file_path = os.path.join(directory_path, filename)
        
        # 处理该CSV文件
        process_csv(file_path)
```

```python
import os
import pandas as pd

# 设置路径
folder_path = r"D:\Python code\oggm\mb-cal"
calibration_file = os.path.join(folder_path, "mb_calibration_results.csv")

# 读取 calibration 文件
df = pd.read_csv(calibration_file)

# 创建新列用于存放 MB 值
df["Multi_Year_Avg_Specific_MB"] = None

# 遍历 RGI_ID
for idx, rgi_id in df["RGI_ID"].items():
    filename = f"mb_ts_{rgi_id}.csv"
    file_path = os.path.join(folder_path, filename)
    if os.path.exists(file_path):
        try:
            temp_df = pd.read_csv(file_path)
            mb_value = temp_df["Multi_Year_Avg_Specific_MB"].iloc[0]
            df.at[idx, "Multi_Year_Avg_Specific_MB"] = mb_value
        except Exception as e:
            print(f"错误读取 {filename}: {e}")

# 保存结果
df.to_csv(calibration_file, index=False)
print("处理完成，已更新文件。")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import mean_absolute_error, mean_squared_error
from scipy.stats import pearsonr
import numpy as np

# 读取 CSV 文件
csv_path = r'D:\Python code\oggm\mb-cal\mb_calibration_results.csv'
df = pd.read_csv(csv_path)

# 提取有效数据（去除缺失值）
df = df[['MB_Geodetic', 'Multi_Year_Avg_Specific_MB']].dropna()

obs = df['MB_Geodetic'].values
sim = df['Multi_Year_Avg_Specific_MB'].values

# --- 1. NSE（Nash–Sutcliffe Efficiency）---
def nse(sim, obs):
    return 1 - np.sum((sim - obs) ** 2) / np.sum((obs - np.mean(obs)) ** 2)

nse_val = nse(sim, obs)

# --- 2. Pearson相关系数 ---
corr, _ = pearsonr(obs, sim)

# --- 3. MAE ---
mae = mean_absolute_error(obs, sim)

# --- 4. RMSE ---
rmse = np.sqrt(mean_squared_error(obs, sim))

# 绘图
plt.figure(figsize=(8, 6))
sns.scatterplot(x=obs, y=sim)
plt.plot([min(obs), max(obs)], [min(obs), max(obs)], 'r--', label='1.1')

plt.xlabel('MB_Geodetic (mm w.e./yr)')
plt.ylabel('Multi-Year Avg Specific MB (mm w.e./yr)')
plt.title('')

# 添加评估指标标注
textstr = '\n'.join((
    f'NSE = {nse_val:.3f}',
    f'Pearson r = {corr:.3f}',
    f'MAE = {mae:.2f}',
    f'RMSE = {rmse:.2f}'
))
plt.text(0.05, 0.95, textstr, transform=plt.gca().transAxes,
         fontsize=10, verticalalignment='top',
         bbox=dict(boxstyle='round,pad=0.4', facecolor='white', edgecolor='gray'))

plt.legend()
plt.grid(True)
plt.tight_layout()

# 导出为 PNG 图片（分辨率 600 dpi）
output_path = r'D:\Python code\oggm\mb-cal\scatter_plot.png'  # 你可以在这里定义你想保存的路径
plt.savefig(output_path, dpi=600)

# 显示图形
plt.show()

print(f'图片已保存为 {output_path}')
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# === 1) 读数据 ===
csv_path = "D:\Python code\oggm\mb-cal\mb_calibration_results.csv"
df = pd.read_csv(csv_path)

# 建议只用成功校准的冰川（更符合“性能评估”语境）
d = df[df["Calibration_Success"] == True].copy()

# === 2) 指标函数 ===
def rmse(a, b):
    a = np.asarray(a); b = np.asarray(b)
    return np.sqrt(np.nanmean((a - b)**2))

def bias(sim, obs):
    sim = np.asarray(sim); obs = np.asarray(obs)
    return np.nanmean(sim - obs)

# === 3) 画布：2×2 ===
fig, axes = plt.subplots(2, 2, figsize=(11, 8.5))

# (a) 校准方法统计
method_counts = d["Calibration_Method"].value_counts()
method_counts.plot(kind="bar", ax=axes[0, 0])
axes[0, 0].set_title("(a) Calibration method")
axes[0, 0].set_ylabel("Number of glaciers")
axes[0, 0].tick_params(axis="x", rotation=30)

# (b) Temp_Bias 分布
tb = d["Temp_Bias"].dropna()
axes[0, 1].hist(tb, bins=25)
axes[0, 1].axvline(0, linestyle="--", linewidth=1)
axes[0, 1].set_title("(b) Temperature bias")
axes[0, 1].set_xlabel("Temp_Bias (°C)")
axes[0, 1].set_ylabel("Count")

# (c) melt_f 分布（箱线图更“冰冻圈”）
mf = d["melt_f"].dropna()
axes[1, 0].boxplot(mf, vert=True, showfliers=True)
axes[1, 0].set_title("(c) Melt factor")
axes[1, 0].set_ylabel("melt_f")
axes[1, 0].set_xticks([])

# (d) 观测-模拟散点图
obs = d["MB_Geodetic"].astype(float)
sim = d["Multi_Year_Avg_Specific_MB"].astype(float)

# 去掉缺失
m = obs.notna() & sim.notna()
obs = obs[m].values
sim = sim[m].values

axes[1, 1].scatter(obs, sim, s=18)

# 1:1 line
mn = np.nanmin([obs.min(), sim.min()])
mx = np.nanmax([obs.max(), sim.max()])
axes[1, 1].plot([mn, mx], [mn, mx], linestyle="--", linewidth=1)

# 拟合线（可选，冰冻圈常见）
if len(obs) >= 2:
    p = np.polyfit(obs, sim, 1)
    xx = np.linspace(mn, mx, 100)
    axes[1, 1].plot(xx, p[0]*xx + p[1], linewidth=1)

# 指标
R = np.corrcoef(obs, sim)[0, 1] if len(obs) > 1 else np.nan
RMSE = rmse(sim, obs)
BIAS = bias(sim, obs)
n = len(obs)

txt = f"n = {n}\nR = {R:.2f}\nRMSE = {RMSE:.1f}\nBias = {BIAS:.1f}"
axes[1, 1].text(0.05, 0.95, txt, transform=axes[1, 1].transAxes,
                va="top")

axes[1, 1].set_title("(d) Geodetic vs modelled MB")
axes[1, 1].set_xlabel("MB_Geodetic (mm w.e. a$^{-1}$)")
axes[1, 1].set_ylabel("Modelled MB (mm w.e. a$^{-1}$)")

plt.tight_layout()
plt.savefig("FigureX_OGGM_calibration_panels.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from mpl_toolkits.axes_grid1.inset_locator import inset_axes

# =========================
# 全局风格：白底、无网格、粗体标签
# =========================
plt.rcParams['font.sans-serif'] = ['Times New Roman', 'Arial']
plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['axes.facecolor'] = 'white'
plt.rcParams['figure.facecolor'] = 'white'

COL_OGGM = "#4b9cd3"   # 蓝
COL_WGMS = "#e57f3f"   # 橙
COL_FIT  = "#E9967A"   # 拟合线/阴影
COL_ZERO = "gray"      # 0线/1:1线
COL_MARGINAL = "#9a9a9a"

# =========================
# 工具函数
# =========================
def rmse(a, b):
    a = np.asarray(a)
    b = np.asarray(b)
    return np.sqrt(np.nanmean((a - b) ** 2))

def bias(sim, obs):
    sim = np.asarray(sim)
    obs = np.asarray(obs)
    return np.nanmean(sim - obs)

def style_axes(ax, ticksize=12):
    ax.grid(False)
    ax.tick_params(axis='both', labelsize=ticksize)
    for s in ax.spines.values():
        s.set_linewidth(1)

def add_panel_label(ax, label, x=-0.1, y=1.05):
    ax.text(
        x, y, label,
        transform=ax.transAxes,
        fontsize=18,
        fontweight='bold',
        va='top',
        ha='right'
    )

def scatter_with_fit(ax, x, y, years,
                     xlabel="", ylabel="", box_text="",
                     cmap="viridis"):
    sc = ax.scatter(
        x, y, c=years, cmap=cmap,
        s=40, alpha=0.7, edgecolors='white', linewidth=0.5, zorder=2
    )

    mn = min(np.nanmin(x), np.nanmin(y))
    mx = max(np.nanmax(x), np.nanmax(y))
    ax.set_xlim(mn, mx)
    ax.set_ylim(mn, mx)
    ax.set_aspect('equal', adjustable='box')

    ax.plot([mn, mx], [mn, mx], color=COL_ZERO, linestyle='--', linewidth=1, zorder=1)

    z = np.polyfit(x, y, 1)
    p = np.poly1d(z)
    xx = np.linspace(mn, mx, 200)
    yy = p(xx)

    rng = np.random.default_rng(42)
    n_boot = 1000
    slopes = np.empty(n_boot)
    intercepts = np.empty(n_boot)

    for i in range(n_boot):
        idx = rng.integers(0, len(x), len(x))
        b = np.polyfit(x[idx], y[idx], 1)
        slopes[i], intercepts[i] = b

    y_boot = np.outer(slopes, xx) + intercepts[:, None]
    y_low, y_high = np.percentile(y_boot, [2.5, 97.5], axis=0)

    ax.fill_between(
        xx, y_low, y_high,
        color=COL_FIT,
        alpha=0.30,
        linewidth=0,
        zorder=2.2
    )
    ax.plot(xx, yy, color=COL_FIT, linestyle=':', linewidth=1.5, zorder=3)

    ax.text(
        0.05, 0.95, box_text,
        transform=ax.transAxes,
        fontsize=14,
        fontweight='bold',
        va='top',
        bbox=dict(boxstyle='round', facecolor='white', alpha=0.8)
    )

    ax.set_xlabel(xlabel, fontsize=14, fontweight='bold')
    ax.set_ylabel(ylabel, fontsize=14, fontweight='bold')

    cbar = plt.colorbar(sc, ax=ax)
    cbar.set_label("Year", fontsize=14, fontweight='bold')

    style_axes(ax)

# =========================
# 读取数据
# =========================
file_path_e = r"D:/oggm/论文数据/wgms/TS.TUYUKSUYSKIY.xlsx"
df_e = pd.read_excel(file_path_e, sheet_name="Sheet1")
data_e = df_e[["year", "oggm", "wgms"]].dropna()

file_path_w = r"D:/oggm/论文数据/wgms/URUMQI GLACIER NO. 1 W-BRANCH.xlsx"
df_w = pd.read_excel(file_path_w, sheet_name="Sheet1")
data_w = df_w[["year", "oggm", "wgms"]].dropna()

csv_path = r"D:\Python code\oggm\mb-cal\mb_calibration_results.csv"
df = pd.read_csv(csv_path)
d = df[df["Calibration_Success"] == True].copy()

method_order = ["geodetic_temp_bias", "geodetic_prcp_fac", "geodetic_default", "scalar"]
method_label_map = {
    "geodetic_temp_bias": "temp",
    "geodetic_prcp_fac": "prcp",
    "geodetic_default": "default",
    "scalar": "scalar"
}
method_counts = (
    d["Calibration_Method"]
    .value_counts()
    .reindex(method_order, fill_value=0)
)

# =========================
# 总图布局
# =========================
fig = plt.figure(figsize=(20, 15))
outer = fig.add_gridspec(2, 1, height_ratios=[1, 1.65], hspace=0.22)

# 顶部
gs_top = outer[0].subgridspec(
    1, 2,
    width_ratios=[1.6, 1.4],
    wspace=0.01
)
axA = fig.add_subplot(gs_top[0, 0])
axC = fig.add_subplot(gs_top[0, 1])

# 底部
gs_bottom = outer[1].subgridspec(
    2, 3,
    width_ratios=[2, 2, 2],
    wspace=0.02,
    hspace=0.35
)
ax1 = fig.add_subplot(gs_bottom[0, 0:2])  # (c)
ax2 = fig.add_subplot(gs_bottom[1, 0:2])  # (e)
ax3 = fig.add_subplot(gs_bottom[0, 2])    # (d)
ax4 = fig.add_subplot(gs_bottom[1, 2])    # (f)

# ==========================================================
# 顶部 (a): 气泡图 + 边缘直方图 + 回归线 + 95%拟合区间 + 1:1线
# ==========================================================
obs = d["MB_Geodetic"].astype(float).values
sim = d["Multi_Year_Avg_Specific_MB"].astype(float).values
m = np.isfinite(obs) & np.isfinite(sim)
obs = obs[m]
sim = sim[m]

# 局部密度 -> 气泡大小
xbins = np.linspace(np.nanmin(obs), np.nanmax(obs), 30)
ybins = np.linspace(np.nanmin(sim), np.nanmax(sim), 30)
H, xedges, yedges = np.histogram2d(obs, sim, bins=[xbins, ybins])

xidx = np.clip(np.digitize(obs, xedges) - 1, 0, H.shape[0] - 1)
yidx = np.clip(np.digitize(sim, yedges) - 1, 0, H.shape[1] - 1)
density = H[xidx, yidx]

# 气泡大小
sizes = 18 + density * 22

# 主图气泡
axA.scatter(
    obs, sim,
    s=sizes,
    facecolor='none',
    edgecolor='coral',
    alpha=0.45,
    linewidth=0.3,
    zorder=2
)

# 1:1线
mn = min(obs.min(), sim.min())
mx = max(obs.max(), sim.max())
axA.plot([mn, mx], [mn, mx], color=COL_ZERO, linestyle='--', linewidth=1, zorder=1)

# 回归线
z = np.polyfit(obs, sim, 1)
xx = np.linspace(mn, mx, 200)
yy = z[0] * xx + z[1]

# bootstrap 95% CI
rng = np.random.default_rng(42)
n_boot = 1000
slopes = np.empty(n_boot)
intercepts = np.empty(n_boot)

for i in range(n_boot):
    idx = rng.integers(0, len(obs), len(obs))
    b = np.polyfit(obs[idx], sim[idx], 1)
    slopes[i], intercepts[i] = b

y_boot = np.outer(slopes, xx) + intercepts[:, None]
y_low, y_high = np.percentile(y_boot, [2.5, 97.5], axis=0)

axA.fill_between(
    xx, y_low, y_high,
    color=COL_FIT,
    alpha=0.26,
    linewidth=0,
    zorder=2.5
)
axA.plot(xx, yy, color=COL_FIT, linestyle=':', linewidth=2, zorder=3)

# 统计值
R = np.corrcoef(obs, sim)[0, 1] if len(obs) > 1 else np.nan
RMSE = rmse(sim, obs)
BIAS = bias(sim, obs)

axA.text(
    0.05, 0.95,
    f"R = {R:.2f}\nRMSE = {RMSE:.1f}\nBias = {BIAS:.1f}",
    transform=axA.transAxes,
    fontsize=13,
    fontweight='bold',
    va='top',
    bbox=dict(boxstyle='round', facecolor='white', alpha=0.85)
)

axA.set_xlabel("MB_Geodetic (mm w.e. a$^{-1}$)", fontsize=16, fontweight='bold')
axA.set_ylabel("Modelled MB (mm w.e. a$^{-1}$)", fontsize=16, fontweight='bold')
axA.set_xlim(mn, mx)
axA.set_ylim(mn, mx)
axA.set_aspect('equal', adjustable='box')
style_axes(axA)

# 气泡大小图例
legend_levels = [1, 3, 6]
legend_sizes = [18 + lv * 22 for lv in legend_levels]
bubble_handles = [
    plt.scatter([], [], s=s, facecolor='none', edgecolor='coral', linewidth=0.5, alpha=0.8)
    for s in legend_sizes
]

leg = axA.legend(
    bubble_handles,
    [f"{lv:.0f}" for lv in legend_levels],
    title="Bubble size\n(density)",
    scatterpoints=1,
    fontsize=10,
    title_fontsize=11,
    frameon=True,
    facecolor='white',
    edgecolor='black',
    loc='lower right',
    bbox_to_anchor=(0.98, 0.04)
)
leg.get_frame().set_alpha(0.9)

# --------------------------
# 上方边缘直方图
# --------------------------
axA_top = inset_axes(
    axA,
    width="78%",
    height="18%",
    loc='upper center',
    bbox_to_anchor=(0.0, 0.10, 1.1, 1.1),
    bbox_transform=axA.transAxes,
    borderpad=0
)

axA_top.hist(
    obs,
    bins=25,
    color=COL_MARGINAL,
    alpha=0.35,
    edgecolor=COL_MARGINAL,
    linewidth=0.6
)
axA_top.set_xlim(mn, mx)
axA_top.set_xticks([])
axA_top.set_yticks([])
for side in ['top', 'right', 'left']:
    axA_top.spines[side].set_visible(False)
axA_top.spines['bottom'].set_linewidth(1)
axA_top.patch.set_alpha(0)
axA_top.grid(False)

# --------------------------
# 右侧边缘直方图
# --------------------------
axA_right = inset_axes(
    axA,
    width="18%",
    height="78%",
    loc='center right',
    bbox_to_anchor=(0.10, 0.02, 1.1, 1.1),
    bbox_transform=axA.transAxes,
    borderpad=0
)

axA_right.hist(
    sim,
    bins=25,
    orientation='horizontal',
    color=COL_MARGINAL,
    alpha=0.35,
    edgecolor=COL_MARGINAL,
    linewidth=0.6
)
axA_right.set_ylim(mn, mx)
axA_right.set_xticks([])
axA_right.set_yticks([])
for side in ['top', 'right', 'bottom']:
    axA_right.spines[side].set_visible(False)
axA_right.spines['left'].set_linewidth(1)
axA_right.patch.set_alpha(0)
axA_right.grid(False)

# ==========================================================
# 顶部右侧主图：原图(c) Temp_Bias
# ==========================================================

tb = d["Temp_Bias"].dropna()
axC.hist(tb, bins=25, color=COL_WGMS, alpha=0.4,rwidth=0.9)
axC.axvline(0, color=COL_ZERO, linestyle='--', linewidth=1)
axC.set_xlabel("Temp_Bias (°C)", fontsize=16, fontweight='bold')
axC.set_ylabel("Count", fontsize=16, fontweight='bold')
style_axes(axC)

# 将原图(b)嵌入到图(c)内部
axB_inset = inset_axes(
    axC,
    width="60%",
    height="42%",
    bbox_to_anchor=(-0.3, -0.1, 1, 1),
    bbox_transform=axC.transAxes,
    borderpad=0
)

x = np.arange(len(method_order))
axB_inset.bar(x, method_counts.values, color=COL_OGGM, alpha=0.9)
axB_inset.set_ylabel("N", fontsize=16, fontweight='bold')
axB_inset.set_xticks(x)
axB_inset.set_xticklabels(
    [method_label_map[k] for k in method_order],
    fontsize=14,
    rotation=0
)
axB_inset.tick_params(axis='y', labelsize=9)
for s in axB_inset.spines.values():
    s.set_linewidth(1)
axB_inset.grid(False)
axB_inset.set_facecolor('white')
axB_inset.set_title("Calibration method", fontsize=16, fontweight='bold', pad=4)

# panel 标签
add_panel_label(axA, "(a)", x=-0.10, y=1.06)
add_panel_label(axC, "(b)", x=-0.04, y=1.05)

# ==========================================================
# 底部：WGMS 四图
# ==========================================================
# (c) E-BRANCH 折线图
ax1.plot(
    data_e["year"], data_e["oggm"],
    marker="s", linewidth=2,
    color=COL_OGGM,
    markerfacecolor='none',
    markeredgecolor=COL_OGGM,
    label="OGGM",
    markersize=8
)
ax1.plot(
    data_e["year"], data_e["wgms"],
    marker="^", linewidth=2.5,
    color=COL_WGMS,
    markerfacecolor='none',
    markeredgecolor=COL_WGMS,
    label="WGMS",
    markersize=8
)
ax1.set_ylabel("Mass Balance (mm w.e.)", fontsize=16, fontweight='bold')
ax1.axhline(0, color=COL_ZERO, linestyle='--', linewidth=1)
ax1.legend(fontsize=14, loc="upper right", framealpha=0.9)
style_axes(ax1)

# (e) W-BRANCH 折线图
ax2.plot(
    data_w["year"], data_w["oggm"],
    marker="s", linewidth=2,
    color=COL_OGGM,
    markerfacecolor='none',
    markeredgecolor=COL_OGGM,
    label="OGGM",
    markersize=8
)
ax2.plot(
    data_w["year"], data_w["wgms"],
    marker="^", linewidth=2.5,
    color=COL_WGMS,
    markerfacecolor='none',
    markeredgecolor=COL_WGMS,
    label="WGMS",
    markersize=8
)
ax2.set_xlabel("Year", fontsize=16, fontweight='bold')
ax2.set_ylabel("Mass Balance (mm w.e.)", fontsize=16, fontweight='bold')
ax2.axhline(0, color=COL_ZERO, linestyle='--', linewidth=1)
ax2.legend(fontsize=14, loc="upper right", framealpha=0.9)
style_axes(ax2)

# (d) E-BRANCH 散点图
corr_e = data_e["oggm"].corr(data_e["wgms"])
r2_e = corr_e ** 2
scatter_with_fit(
    ax3,
    data_e["oggm"].values,
    data_e["wgms"].values,
    data_e["year"].values,
    xlabel="OGGM (mm w.e.)",
    ylabel="WGMS (mm w.e.)",
    box_text=f"Pearson r = {corr_e:.3f}\nR² = {r2_e:.3f}",
    cmap="viridis"
)

# (f) W-BRANCH 散点图
corr_w = data_w["oggm"].corr(data_w["wgms"])
r2_w = corr_w ** 2
scatter_with_fit(
    ax4,
    data_w["oggm"].values,
    data_w["wgms"].values,
    data_w["year"].values,
    xlabel="OGGM (mm w.e.)",
    ylabel="WGMS (mm w.e.)",
    box_text=f"Pearson r = {corr_w:.3f}\nR² = {r2_w:.3f}",
    cmap="viridis"
)

# 底部 panel 标号
add_panel_label(ax1, "(c)")
add_panel_label(ax3, "(d)")
add_panel_label(ax2, "(e)")
add_panel_label(ax4, "(f)")

# =========================
# 保存
# =========================
out_path = r"D:/oggm/论文数据/wgms/Figure_Combined_Final.png"
plt.savefig(out_path, dpi=300, bbox_inches="tight", facecolor='white')
plt.show()

print("TS.TUYUKSUYSKIY.xlsx:")
print(f"Pearson r: {corr_e:.3f}")
print(f"R-squared: {r2_e:.3f}")

print("\nW-BRANCH Statistics:")
print(f"Pearson r: {corr_w:.3f}")
print(f"R-squared: {r2_w:.3f}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.axes_grid1.inset_locator import inset_axes
from matplotlib.ticker import MaxNLocator
from scipy import stats

# ==========================================================
# 全局风格
# ==========================================================
plt.rcParams.update({
    "font.family": "Times New Roman",
    "mathtext.fontset": "stix",
    "axes.unicode_minus": False,
    "axes.facecolor": "white",
    "figure.facecolor": "white",
    "axes.linewidth": 1.0,
    "xtick.direction": "out",
    "ytick.direction": "out",
    "xtick.major.size": 4,
    "ytick.major.size": 4,
    "xtick.major.width": 0.9,
    "ytick.major.width": 0.9
})

# ==========================================================
# 配色
# ==========================================================
COL_OGGM = "#3F7FBF"
COL_WGMS = "#D97835"
COL_FIT = "#A9502D"
COL_ZERO = "#666666"
COL_TEXT = "#222222"
COL_SPINE = "#333333"
COL_METHODS = ["#4C78A8", "#72B7B2", "#F2B447", "#D65F5F"]
COL_BUBBLE = "#8E44AD"  # 气泡图紫色，避免与直方图冲突

# ==========================================================
# 工具函数
# ==========================================================
def rmse(a, b):
    a, b = np.asarray(a, dtype=float), np.asarray(b, dtype=float)
    return np.sqrt(np.nanmean((a - b) ** 2))

def bias(sim, obs):
    sim, obs = np.asarray(sim, dtype=float), np.asarray(obs, dtype=float)
    return np.nanmean(sim - obs)

def style_axes(ax, ticksize=12):
    ax.grid(False)
    ax.tick_params(axis="both", labelsize=ticksize, colors=COL_TEXT)
    for spine in ax.spines.values():
        spine.set_linewidth(1.0)
        spine.set_color(COL_SPINE)

def add_panel_label(ax, label, x=-0.09, y=1.045):
    ax.text(x, y, label, transform=ax.transAxes, fontsize=17, fontweight="bold",
            va="top", ha="right", color=COL_TEXT, clip_on=False)

def add_statistics_box(ax, text, x=0.045, y=0.955, fontsize=12):
    ax.text(x, y, text, transform=ax.transAxes, fontsize=fontsize, fontweight="bold",
            va="top", ha="left", color=COL_TEXT,
            bbox=dict(boxstyle="round,pad=0.28", facecolor="white",
                      edgecolor="#B0B0B0", linewidth=0.65, alpha=0.88))

def bootstrap_fit_ci(x, y, xx, n_boot=1000, seed=42):
    x, y = np.asarray(x, dtype=float), np.asarray(y, dtype=float)
    rng = np.random.default_rng(seed)
    boot_predictions = np.empty((n_boot, len(xx)))

    for i in range(n_boot):
        idx = rng.integers(0, len(x), len(x))
        coef = np.polyfit(x[idx], y[idx], 1)
        boot_predictions[i] = np.polyval(coef, xx)

    return np.percentile(boot_predictions, [2.5, 97.5], axis=0)

def scatter_with_fit(ax, x, y, years, xlabel="", ylabel="", box_text="", cmap="cividis"):
    x, y, years = np.asarray(x, dtype=float), np.asarray(y, dtype=float), np.asarray(years)

    sc = ax.scatter(x, y, c=years, cmap=cmap, s=42, alpha=0.82,
                    edgecolors="white", linewidth=0.55, zorder=3)

    mn, mx = min(np.nanmin(x), np.nanmin(y)), max(np.nanmax(x), np.nanmax(y))
    padding = (mx - mn) * 0.04 if mx > mn else 1
    plot_min, plot_max = mn - padding, mx + padding

    ax.plot([plot_min, plot_max], [plot_min, plot_max], color=COL_ZERO,
            linestyle="--", linewidth=1.1, zorder=1)

    coef = np.polyfit(x, y, 1)
    xx = np.linspace(plot_min, plot_max, 200)
    yy = np.polyval(coef, xx)
    y_low, y_high = bootstrap_fit_ci(x, y, xx)

    ax.fill_between(xx, y_low, y_high, color=COL_FIT, alpha=0.14, linewidth=0, zorder=1.5)
    ax.plot(xx, yy, color=COL_FIT, linestyle=":", linewidth=1.8, zorder=2)

    add_statistics_box(ax, box_text, fontsize=11)
    ax.set_xlabel(xlabel, fontsize=14, fontweight="bold")
    ax.set_ylabel(ylabel, fontsize=14, fontweight="bold")
    ax.set_xlim(plot_min, plot_max)
    ax.set_ylim(plot_min, plot_max)
    ax.set_aspect("equal", adjustable="box")
    style_axes(ax, ticksize=11)

    cbar = plt.colorbar(sc, ax=ax, fraction=0.046, pad=0.035)
    cbar.set_label("Year", fontsize=12, fontweight="bold")
    cbar.ax.tick_params(labelsize=10)
    cbar.outline.set_linewidth(0.8)
    return sc

# ==========================================================
# 读取数据
# ==========================================================
file_path_e = r"D:/oggm/论文数据/wgms/TS.TUYUKSUYSKIY.xlsx"
file_path_w = r"D:/oggm/论文数据/wgms/URUMQI GLACIER NO. 1 W-BRANCH.xlsx"
csv_path = r"D:\Python code\oggm\mb-cal\mb_calibration_results.csv"

df_e = pd.read_excel(file_path_e, sheet_name="Sheet1")
df_w = pd.read_excel(file_path_w, sheet_name="Sheet1")
df = pd.read_csv(csv_path)

data_e = df_e[["year", "oggm", "wgms"]].dropna().sort_values("year")
data_w = df_w[["year", "oggm", "wgms"]].dropna().sort_values("year")
d = df.loc[df["Calibration_Success"].eq(True)].copy()

method_order = ["geodetic_temp_bias", "geodetic_prcp_fac", "geodetic_default", "scalar"]
method_label_map = {
    "geodetic_temp_bias": "Temp.",
    "geodetic_prcp_fac": "Prcp.",
    "geodetic_default": "Default",
    "scalar": "Scalar"
}
method_counts = d["Calibration_Method"].value_counts().reindex(method_order, fill_value=0)

# ==========================================================
# 总图布局 - 紧凑间距
# ==========================================================
fig = plt.figure(figsize=(14, 14))
outer = fig.add_gridspec(2, 1, height_ratios=[1.02, 1.58], hspace=0.18)  # 减小上下间距

gs_top = outer[0].subgridspec(1, 2, width_ratios=[1.34, 1.0], wspace=0.01)  # 减小左右间距
axA = fig.add_subplot(gs_top[0, 0])
axB = fig.add_subplot(gs_top[0, 1])

gs_bottom = outer[1].subgridspec(
    2, 2, width_ratios=[1.4, 1.0], height_ratios=[1, 1],
    wspace=0.01, hspace=0.2  # 大幅减小间距
)
axC = fig.add_subplot(gs_bottom[0, 0])
axD = fig.add_subplot(gs_bottom[0, 1])
axE = fig.add_subplot(gs_bottom[1, 0])
axF = fig.add_subplot(gs_bottom[1, 1])

# ==========================================================
# (a) Geodetic MB 与 OGGM模拟值 - 紫色气泡
# ==========================================================
obs = pd.to_numeric(d["MB_Geodetic"], errors="coerce").to_numpy()
sim = pd.to_numeric(d["Multi_Year_Avg_Specific_MB"], errors="coerce").to_numpy()
mask = np.isfinite(obs) & np.isfinite(sim)
obs, sim = obs[mask], sim[mask]

xbins = np.linspace(obs.min(), obs.max(), 30)
ybins = np.linspace(sim.min(), sim.max(), 30)
H, xedges, yedges = np.histogram2d(obs, sim, bins=[xbins, ybins])
xidx = np.clip(np.digitize(obs, xedges) - 1, 0, H.shape[0] - 1)
yidx = np.clip(np.digitize(sim, yedges) - 1, 0, H.shape[1] - 1)
density = H[xidx, yidx]
sizes = 18 + density * 22

# 使用紫色气泡，避免与直方图颜色冲突
axA.scatter(obs, sim, s=sizes, facecolor="none", edgecolor=COL_BUBBLE,
            alpha=0.55, linewidth=0.6, zorder=3)

mn, mx = min(obs.min(), sim.min()), max(obs.max(), sim.max())
pad = (mx - mn) * 0.025
mn_plot, mx_plot = mn - pad, mx + pad

axA.plot([mn_plot, mx_plot], [mn_plot, mx_plot], color=COL_ZERO,
         linestyle="--", linewidth=1.15, zorder=1)

coef = np.polyfit(obs, sim, 1)
xx = np.linspace(mn_plot, mx_plot, 250)
yy = np.polyval(coef, xx)
y_low, y_high = bootstrap_fit_ci(obs, sim, xx)

axA.fill_between(xx, y_low, y_high, color=COL_FIT, alpha=0.17, linewidth=0, zorder=1.5)
axA.plot(xx, yy, color=COL_FIT, linestyle=":", linewidth=2.0, zorder=2)

R = np.corrcoef(obs, sim)[0, 1] if len(obs) > 1 else np.nan
RMSE = rmse(sim, obs)
BIAS = bias(sim, obs)

add_statistics_box(axA, f"$R$ = {R:.2f}\nRMSE = {RMSE:.1f}\nBias = {BIAS:.1f}",
                   x=0.045, y=0.95, fontsize=12)

axA.set_xlabel("Geodetic MB (mm w.e. a$^{-1}$)", fontsize=15, fontweight="bold")
axA.set_ylabel("Modelled MB (mm w.e. a$^{-1}$)", fontsize=15, fontweight="bold")
axA.set_xlim(mn_plot, mx_plot)
axA.set_ylim(mn_plot, mx_plot)
axA.set_aspect("equal", adjustable="box")
style_axes(axA, ticksize=12)

# 气泡密度图例（紫色）
legend_levels = [1, 3, 6]
bubble_handles = [
    axA.scatter([], [], s=18 + lv * 22, facecolor="none", edgecolor=COL_BUBBLE,
                linewidth=0.7, alpha=0.8) for lv in legend_levels
]
axA.legend(
    bubble_handles, [str(lv) for lv in legend_levels],
    title="Point density", scatterpoints=1,
    fontsize=9.5, title_fontsize=10.5,
    frameon=False, loc="lower right",
    bbox_to_anchor=(0.985, 0.025),
    borderpad=0.1, labelspacing=0.55, handletextpad=0.7
)

# 【修复】上方边缘直方图 - 使用橙色与Geodetic对应
axA_top = axA.inset_axes([0, 1.0, 1, 0.15])  # x, y, width, height (相对于axA)
axA_top.hist(obs, bins=25, color=COL_WGMS, alpha=0.30,
             edgecolor=COL_WGMS, linewidth=0.65, rwidth=0.93)
axA_top.set_xlim(mn_plot, mx_plot)
axA_top.set_xticks([])
axA_top.set_yticks([])
axA_top.patch.set_alpha(0)
for side in ["top", "right", "left"]:
    axA_top.spines[side].set_visible(False)
axA_top.spines["bottom"].set_color(COL_SPINE)
axA_top.spines["bottom"].set_linewidth(0.8)

# 【修复】右侧边缘直方图 - 使用蓝色与OGGM对应
axA_right = axA.inset_axes([1.0, 0, 0.15, 1])
axA_right.hist(sim, bins=25, orientation="horizontal", color=COL_OGGM,
               alpha=0.30, edgecolor=COL_OGGM, linewidth=0.65, rwidth=0.93)
axA_right.set_ylim(mn_plot, mx_plot)
axA_right.set_xticks([])
axA_right.set_yticks([])
axA_right.patch.set_alpha(0)
for side in ["top", "right", "bottom"]:
    axA_right.spines[side].set_visible(False)
axA_right.spines["left"].set_color(COL_SPINE)
axA_right.spines["left"].set_linewidth(0.8)

add_panel_label(axA, "(a)", x=-0.085, y=1.055)

# ==========================================================
# (b) Temp_Bias + 嵌入柱状图（恢复原始数据）
# ==========================================================
tb = pd.to_numeric(d["Temp_Bias"], errors="coerce").dropna()

counts, bins, patches = axB.hist(
    tb, bins=25, color=COL_OGGM, alpha=0.27,
    edgecolor=COL_OGGM, linewidth=0.9, rwidth=0.90
)

tb_mean, tb_median = tb.mean(), tb.median()
axB.axvline(0, color="#444444", linestyle="--", linewidth=1.15, zorder=4)
axB.axvline(tb_mean, color=COL_WGMS, linestyle="-", linewidth=1.65, zorder=4)
axB.axvline(tb_median, color=COL_FIT, linestyle=":", linewidth=1.65, zorder=4)

axB.set_xlabel("Temperature bias (°C)", fontsize=15, fontweight="bold")
axB.set_ylabel("Number of glaciers", fontsize=15, fontweight="bold")
axB.yaxis.set_major_locator(MaxNLocator(integer=True, nbins=6))
style_axes(axB, ticksize=12)

axB.text(
    0.045, 0.95,
    f"Mean = {tb_mean:.2f} °C\nMedian = {tb_median:.2f} °C\n$n$ = {len(tb)}",
    transform=axB.transAxes, fontsize=11, fontweight="bold",
    va="top", ha="left",
    bbox=dict(boxstyle="round,pad=0.28", facecolor="white",
              edgecolor="#B0B0B0", linewidth=0.65, alpha=0.88)
)

reference_handles = [
    plt.Line2D([0], [0], color=COL_WGMS, linewidth=1.7, linestyle="-", label="Mean"),
    plt.Line2D([0], [0], color=COL_FIT, linewidth=1.7, linestyle=":", label="Median"),
    plt.Line2D([0], [0], color="#444444", linewidth=1.2, linestyle="--", label="Zero")
]
axB.legend(handles=reference_handles, frameon=False, fontsize=9.5,
           loc="upper left", bbox_to_anchor=(0.03, 0.69),
           handlelength=2.3, labelspacing=0.35)

# 嵌入图：恢复为Calibration schemes柱状图
axB_inset = axB.inset_axes([0.2, 0.4, 0.37, 0.7])  # 右上角

x_method = np.arange(len(method_order))
bars = axB_inset.bar(
    x_method, method_counts.values,
    width=0.66, color=COL_METHODS,
    edgecolor="white", linewidth=0.7, alpha=0.95, zorder=2
)

axB_inset.set_title("Calibration schemes", fontsize=10.5, fontweight="bold", pad=5)
axB_inset.set_ylabel("Count", fontsize=9.5, fontweight="bold")
axB_inset.set_xticks(x_method)
axB_inset.set_xticklabels([method_label_map[k] for k in method_order],
                          fontsize=10.5, rotation=20, ha="right")
axB_inset.tick_params(axis="y", labelsize=8.5, length=2.5, width=0.7)
axB_inset.tick_params(axis="x", length=0)
axB_inset.yaxis.set_major_locator(MaxNLocator(integer=True, nbins=4))
axB_inset.grid(axis="y", color="#D8D8D8", linewidth=0.55, alpha=0.55, zorder=0)
axB_inset.set_facecolor("white")

for spine in axB_inset.spines.values():
    spine.set_linewidth(0.8)
    spine.set_color("#777777")

offset = max(method_counts.max() * 0.025, 0.5)
for bar, value in zip(bars, method_counts.values):
    axB_inset.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + offset,
                   f"{int(value)}", ha="center", va="bottom",
                   fontsize=8.5, fontweight="bold", color=COL_TEXT)

axB_inset.set_ylim(0, method_counts.max() * 1.18 if method_counts.max() > 0 else 1)
add_panel_label(axB, "(b)", x=-0.07, y=1.055)

# ==========================================================
# (c) Tuyuksuyskiy Glacier 年际质量平衡
# ==========================================================
axC.plot(data_e["year"], data_e["oggm"], marker="s", linewidth=2.0,
         color=COL_OGGM, markerfacecolor="none", markeredgecolor=COL_OGGM,
         markeredgewidth=1.25, label="OGGM", markersize=7.5)

axC.plot(data_e["year"], data_e["wgms"], marker="^", linewidth=2.3,
         color=COL_WGMS, markerfacecolor="none", markeredgecolor=COL_WGMS,
         markeredgewidth=1.25, label="WGMS", markersize=7.5)

axC.axhline(0, color=COL_ZERO, linestyle="--", linewidth=1)
axC.set_ylabel("Mass balance (mm w.e.)", fontsize=15, fontweight="bold")
axC.legend(fontsize=12, loc="upper right", frameon=False, ncol=2,
           handlelength=2.2, columnspacing=1.2)
style_axes(axC, ticksize=12)
add_panel_label(axC, "(c)", x=-0.07, y=1.045)

# ==========================================================
# (d) Tuyuksuyskiy Glacier 相关性
# ==========================================================
corr_e = data_e["oggm"].corr(data_e["wgms"])
r2_e = corr_e ** 2

scatter_with_fit(
    axD,
    data_e["oggm"].values,
    data_e["wgms"].values,
    data_e["year"].values,
    xlabel="OGGM (mm w.e.)",
    ylabel="WGMS (mm w.e.)",
    box_text=f"Pearson $r$ = {corr_e:.3f}\n$R^2$ = {r2_e:.3f}",
    cmap="cividis"
)
add_panel_label(axD, "(d)", x=-0.12, y=1.045)

# ==========================================================
# (e) Urumqi Glacier No. 1 West Branch 年际质量平衡
# ==========================================================
axE.plot(data_w["year"], data_w["oggm"], marker="s", linewidth=2.0,
         color=COL_OGGM, markerfacecolor="none", markeredgecolor=COL_OGGM,
         markeredgewidth=1.25, label="OGGM", markersize=7.5)

axE.plot(data_w["year"], data_w["wgms"], marker="^", linewidth=2.3,
         color=COL_WGMS, markerfacecolor="none", markeredgecolor=COL_WGMS,
         markeredgewidth=1.25, label="WGMS", markersize=7.5)

axE.axhline(0, color=COL_ZERO, linestyle="--", linewidth=1)
axE.set_xlabel("Year", fontsize=15, fontweight="bold")
axE.set_ylabel("Mass balance (mm w.e.)", fontsize=15, fontweight="bold")
axE.legend(fontsize=12, loc="upper right", frameon=False, ncol=2,
           handlelength=2.2, columnspacing=1.2)
style_axes(axE, ticksize=12)
add_panel_label(axE, "(e)", x=-0.07, y=1.045)

# ==========================================================
# (f) Urumqi Glacier No. 1 West Branch 相关性
# ==========================================================
corr_w = data_w["oggm"].corr(data_w["wgms"])
r2_w = corr_w ** 2

scatter_with_fit(
    axF,
    data_w["oggm"].values,
    data_w["wgms"].values,
    data_w["year"].values,
    xlabel="OGGM (mm w.e.)",
    ylabel="WGMS (mm w.e.)",
    box_text=f"Pearson $r$ = {corr_w:.3f}\n$R^2$ = {r2_w:.3f}",
    cmap="cividis"
)
add_panel_label(axF, "(f)", x=-0.12, y=1.045)

# ==========================================================
# 统一调整
# ==========================================================
for ax in [axA, axB, axC, axD, axE, axF]:
    ax.set_axisbelow(True)

fig.align_ylabels([axA, axC, axE])
fig.align_ylabels([axB, axD, axF])

# ==========================================================
# 保存
# ==========================================================
out_path = r"D:/oggm/论文数据/wgms/Figure_Combined_Optimized.png"

plt.savefig(
    out_path,
    dpi=600,
    bbox_inches="tight",
    pad_inches=0.04,  # 减小边距
    facecolor="white"
)
plt.show()

# ==========================================================
# 输出统计结果
# ==========================================================
print("Figure saved to:")
print(out_path)

print("\nGeodetic calibration statistics:")
print(f"R: {R:.3f}")
print(f"RMSE: {RMSE:.3f}")
print(f"Bias: {BIAS:.3f}")

print("\nTuyuksuyskiy Glacier:")
print(f"Pearson r: {corr_e:.3f}")
print(f"R-squared: {r2_e:.3f}")

print("\nUrumqi Glacier No. 1 West Branch:")
print(f"Pearson r: {corr_w:.3f}")
print(f"R-squared: {r2_w:.3f}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import MaxNLocator

# ==========================================================
# 全局风格
# ==========================================================
plt.rcParams.update({
    "font.family": "Times New Roman",
    "mathtext.fontset": "stix",
    "axes.unicode_minus": False,
    "axes.facecolor": "white",
    "figure.facecolor": "white",
    "axes.linewidth": 1.0,
    "xtick.direction": "out",
    "ytick.direction": "out",
    "xtick.major.size": 4,
    "ytick.major.size": 4,
    "xtick.major.width": 0.9,
    "ytick.major.width": 0.9
})

# ==========================================================
# 配色
# ==========================================================
COL_OGGM = "#3F7FBF"
COL_WGMS = "#D97835"
COL_FIT = "#A9502D"
COL_ZERO = "#D8BFD8"
COL_TEXT = "#222222"
COL_SPINE = "#333333"
COL_METHODS = ["#72B7B2", "#4C78A8", "#F2B447", "#D65F5F"]
COL_BUBBLE = "#D8BFD8"

# ==========================================================
# 工具函数
# ==========================================================
def rmse(a, b):
    a, b = np.asarray(a, dtype=float), np.asarray(b, dtype=float)
    return np.sqrt(np.nanmean((a - b) ** 2))

def bias(sim, obs):
    sim, obs = np.asarray(sim, dtype=float), np.asarray(obs, dtype=float)
    return np.nanmean(sim - obs)

def style_axes(ax, ticksize=12):
    ax.grid(False)
    ax.tick_params(axis="both", labelsize=ticksize, colors=COL_TEXT)
    for spine in ax.spines.values():
        spine.set_linewidth(1.0)
        spine.set_color(COL_SPINE)

def add_panel_label(ax, label, x=-0.09, y=1.045):
    ax.text(x, y, label, transform=ax.transAxes, fontsize=17, fontweight="bold",
            va="top", ha="right", color=COL_TEXT, clip_on=False)

def add_statistics_box(ax, text, x=0.045, y=0.955, fontsize=12):
    ax.text(x, y, text, transform=ax.transAxes, fontsize=fontsize, fontweight="bold",
            va="top", ha="left", color=COL_TEXT,
            bbox=dict(boxstyle="round,pad=0.28", facecolor="white",
                      edgecolor="#B0B0B0", linewidth=0.65, alpha=0.88))

def bootstrap_fit_ci(x, y, xx, n_boot=1000, seed=42):
    x, y = np.asarray(x, dtype=float), np.asarray(y, dtype=float)
    rng = np.random.default_rng(seed)
    boot_predictions = np.empty((n_boot, len(xx)))

    for i in range(n_boot):
        idx = rng.integers(0, len(x), len(x))
        coef = np.polyfit(x[idx], y[idx], 1)
        boot_predictions[i] = np.polyval(coef, xx)

    return np.percentile(boot_predictions, [2.5, 97.5], axis=0)

def scatter_with_fit(ax, x, y, years, xlabel="", ylabel="", box_text="", cmap="cividis"):
    x, y, years = np.asarray(x, dtype=float), np.asarray(y, dtype=float), np.asarray(years)

    sc = ax.scatter(x, y, c=years, cmap=cmap, s=42, alpha=0.82,
                    edgecolors="white", linewidth=0.55, zorder=3)

    mn, mx = min(np.nanmin(x), np.nanmin(y)), max(np.nanmax(x), np.nanmax(y))
    padding = (mx - mn) * 0.04 if mx > mn else 1
    plot_min, plot_max = mn - padding, mx + padding

    ax.plot([plot_min, plot_max], [plot_min, plot_max], color=COL_ZERO,
            linestyle="--", linewidth=1.1, zorder=1)

    coef = np.polyfit(x, y, 1)
    xx = np.linspace(plot_min, plot_max, 200)
    yy = np.polyval(coef, xx)
    y_low, y_high = bootstrap_fit_ci(x, y, xx)

    ax.fill_between(xx, y_low, y_high, color=COL_FIT, alpha=0.14, linewidth=0, zorder=1.5)
    ax.plot(xx, yy, color=COL_FIT, linestyle=":", linewidth=1.8, zorder=2)

    add_statistics_box(ax, box_text, fontsize=11)
    ax.set_xlabel(xlabel, fontsize=14, fontweight="bold")
    ax.set_ylabel(ylabel, fontsize=14, fontweight="bold")
    ax.set_xlim(plot_min, plot_max)
    ax.set_ylim(plot_min, plot_max)
    ax.set_aspect("equal", adjustable="box")
    style_axes(ax, ticksize=11)

    cbar = plt.colorbar(sc, ax=ax, fraction=0.046, pad=0.035)
    cbar.set_label("Year", fontsize=12, fontweight="bold")
    cbar.ax.tick_params(labelsize=10)
    cbar.outline.set_linewidth(0.8)
    return sc

# ==========================================================
# 读取数据
# ==========================================================
file_path_e = r"D:/oggm/论文数据/wgms/TS.TUYUKSUYSKIY.xlsx"
file_path_w = r"D:/oggm/论文数据/wgms/URUMQI GLACIER NO. 1 W-BRANCH.xlsx"
csv_path = r"D:\Python code\oggm\mb-cal\mb_calibration_results.csv"

df_e = pd.read_excel(file_path_e, sheet_name="Sheet1")
df_w = pd.read_excel(file_path_w, sheet_name="Sheet1")
df = pd.read_csv(csv_path)

data_e = df_e[["year", "oggm", "wgms"]].dropna().sort_values("year")
data_w = df_w[["year", "oggm", "wgms"]].dropna().sort_values("year")
d = df.loc[df["Calibration_Success"].eq(True)].copy()

method_order = ["geodetic_temp_bias", "geodetic_prcp_fac", "geodetic_default", "scalar"]
method_label_map = {
    "geodetic_temp_bias": "Temp.",
    "geodetic_prcp_fac": "Prcp.",
    "geodetic_default": "Default",
    "scalar": "Scalar"
}
method_counts = d["Calibration_Method"].value_counts().reindex(method_order, fill_value=0)

# ==========================================================
# 总图布局
# ==========================================================
fig = plt.figure(figsize=(13, 14))
outer = fig.add_gridspec(2, 1, height_ratios=[1.02, 1.58], hspace=0.15)

gs_top = outer[0].subgridspec(1, 2, width_ratios=[1.34, 1.0], wspace=0.08)
axA = fig.add_subplot(gs_top[0, 0])
axB = fig.add_subplot(gs_top[0, 1])

gs_bottom = outer[1].subgridspec(
    2, 2, width_ratios=[1.6, 1.0], height_ratios=[1, 1],
    wspace=0.06, hspace=0.22
)
axC = fig.add_subplot(gs_bottom[0, 0])
axD = fig.add_subplot(gs_bottom[0, 1])
axE = fig.add_subplot(gs_bottom[1, 0])
axF = fig.add_subplot(gs_bottom[1, 1])

# ==========================================================
# (a) Geodetic MB 与 OGGM模拟值
# ==========================================================
obs = pd.to_numeric(d["MB_Geodetic"], errors="coerce").to_numpy()
sim = pd.to_numeric(d["Multi_Year_Avg_Specific_MB"], errors="coerce").to_numpy()
mask = np.isfinite(obs) & np.isfinite(sim)
obs, sim = obs[mask], sim[mask]

xbins = np.linspace(obs.min(), obs.max(), 30)
ybins = np.linspace(sim.min(), sim.max(), 30)
H, xedges, yedges = np.histogram2d(obs, sim, bins=[xbins, ybins])
xidx = np.clip(np.digitize(obs, xedges) - 1, 0, H.shape[0] - 1)
yidx = np.clip(np.digitize(sim, yedges) - 1, 0, H.shape[1] - 1)
density = H[xidx, yidx]
sizes = 18 + density * 22

axA.scatter(obs, sim, s=sizes, facecolor="none", edgecolor=COL_BUBBLE,
            alpha=0.55, linewidth=0.6, zorder=3)

mn, mx = min(obs.min(), sim.min()), max(obs.max(), sim.max())
pad = (mx - mn) * 0.025
mn_plot, mx_plot = mn - pad, mx + pad

axA.plot([mn_plot, mx_plot], [mn_plot, mx_plot], color=COL_ZERO,
         linestyle="--", linewidth=1.15, zorder=1)

coef = np.polyfit(obs, sim, 1)
xx = np.linspace(mn_plot, mx_plot, 250)
yy = np.polyval(coef, xx)
y_low, y_high = bootstrap_fit_ci(obs, sim, xx)

axA.fill_between(xx, y_low, y_high, color=COL_FIT, alpha=0.17, linewidth=0, zorder=1.5)
axA.plot(xx, yy, color=COL_FIT, linestyle=":", linewidth=2.0, zorder=2)

R = np.corrcoef(obs, sim)[0, 1] if len(obs) > 1 else np.nan
RMSE = rmse(sim, obs)
BIAS = bias(sim, obs)

add_statistics_box(axA, f"$R$ = {R:.2f}\nRMSE = {RMSE:.1f}\nBias = {BIAS:.1f}",
                   x=0.045, y=0.95, fontsize=12)

axA.set_xlabel("Geodetic MB (mm w.e.)", fontsize=15, fontweight="bold")
axA.set_ylabel("Modelled MB (mm w.e.)", fontsize=15, fontweight="bold")
axA.set_xlim(mn_plot, mx_plot)
axA.set_ylim(mn_plot, mx_plot)
axA.set_aspect("equal", adjustable="box")
style_axes(axA, ticksize=12)

# 气泡密度图例
legend_levels = [1, 3, 6]
bubble_handles = [
    axA.scatter([], [], s=18 + lv * 22, facecolor="none", edgecolor=COL_BUBBLE,
                linewidth=0.7, alpha=0.8) for lv in legend_levels
]
axA.legend(
    bubble_handles, [str(lv) for lv in legend_levels],
    title="Point density", scatterpoints=1,
    fontsize=9.5, title_fontsize=10.5,
    frameon=False, loc="lower right",
    bbox_to_anchor=(0.985, 0.025),
    borderpad=0.1, labelspacing=0.55, handletextpad=0.7
)

# 【修复】上方边缘直方图 - 紧贴主轴上方，无间隙
axA_top = axA.inset_axes([0, 1.0, 1, 0.15], transform=axA.transAxes)
axA_top.hist(obs, bins=25, color=COL_WGMS, alpha=0.30,
             edgecolor=COL_WGMS, linewidth=0.65, rwidth=0.93)
axA_top.set_xlim(mn_plot, mx_plot)
axA_top.set_xticks([])
axA_top.set_yticks([])
axA_top.patch.set_alpha(0)
for side in ["top", "right", "left"]:
    axA_top.spines[side].set_visible(False)
axA_top.spines["bottom"].set_color(COL_SPINE)
axA_top.spines["bottom"].set_linewidth(0.8)

# 【修复】右侧边缘直方图 - 紧贴主轴右侧，无间隙
axA_right = axA.inset_axes([1.0, 0, 0.15, 1], transform=axA.transAxes)
axA_right.hist(sim, bins=25, orientation="horizontal", color=COL_OGGM,
               alpha=0.30, edgecolor=COL_OGGM, linewidth=0.65, rwidth=0.93)
axA_right.set_ylim(mn_plot, mx_plot)
axA_right.set_xticks([])
axA_right.set_yticks([])
axA_right.patch.set_alpha(0)
for side in ["top", "right", "bottom"]:
    axA_right.spines[side].set_visible(False)
axA_right.spines["left"].set_color(COL_SPINE)
axA_right.spines["left"].set_linewidth(0.8)

# 【修复】确保y轴标签不被遮挡，调整标签位置
axA.yaxis.set_label_coords(-0.12, 0.5)  # 手动调整y轴标签位置

add_panel_label(axA, "(a)", x=-0.085, y=1.055)

# ==========================================================
# (b) Temp_Bias + 中部又窄又高的横向柱状嵌入图
# ==========================================================
tb = pd.to_numeric(d["Temp_Bias"], errors="coerce").dropna()

counts, bins, patches = axB.hist(
    tb, bins=25, color=COL_OGGM, alpha=0.27,
    edgecolor=COL_OGGM, linewidth=0.9, rwidth=0.90
)

tb_mean, tb_median = tb.mean(), tb.median()
axB.axvline(0, color="#444444", linestyle="--", linewidth=1.15, zorder=4)
axB.axvline(tb_mean, color=COL_WGMS, linestyle="-", linewidth=1.65, zorder=4)
axB.axvline(tb_median, color=COL_FIT, linestyle=":", linewidth=1.65, zorder=4)

axB.set_xlabel("Temperature bias (°C)", fontsize=15, fontweight="bold")
axB.set_ylabel("Number of glaciers", fontsize=15, fontweight="bold")
axB.yaxis.set_major_locator(MaxNLocator(integer=True, nbins=6))
style_axes(axB, ticksize=12)

axB.text(
    0.045, 0.95,
    f"Mean = {tb_mean:.2f} °C\nMedian = {tb_median:.2f} °C\n$n$ = {len(tb)}",
    transform=axB.transAxes, fontsize=11, fontweight="bold",
    va="top", ha="left",
)

reference_handles = [
    plt.Line2D([0], [0], color=COL_WGMS, linewidth=1.7, linestyle="-", label="Mean"),
    plt.Line2D([0], [0], color=COL_FIT, linewidth=1.7, linestyle=":", label="Median"),
    plt.Line2D([0], [0], color="#444444", linewidth=1.2, linestyle="--", label="Zero")
]
axB.legend(handles=reference_handles, frameon=False, fontsize=10,
           loc="upper left", bbox_to_anchor=(0.03, 0.69),
           handlelength=2.3, labelspacing=0.35)

# 【修改】中部又窄又高的横向柱状嵌入图
# left=0.35: 居中偏左, bottom=0.15: 底部留空, width=0.3: 窄, height=0.65: 高
axB_inset = axB.inset_axes([0.35, 0.45, 0.3, 0.45], transform=axB.transAxes)

# 横向柱状图：使用barh
x_method = np.arange(len(method_order))
bars = axB_inset.barh(
    x_method, method_counts.values,
    height=0.55, color=COL_METHODS,
    edgecolor="white", linewidth=0.7, alpha=0.95, zorder=2
)

axB_inset.set_title("Calibration\nschemes", fontsize=12, fontweight="bold", pad=6)
axB_inset.set_xlabel("Count", fontsize=12, fontweight="bold")
axB_inset.set_yticks(x_method)
axB_inset.set_yticklabels([method_label_map[k] for k in method_order],
                          fontsize=12)
axB_inset.tick_params(axis="x", labelsize=12, length=2.5, width=0.7)
axB_inset.tick_params(axis="y", length=0)
axB_inset.xaxis.set_major_locator(MaxNLocator(integer=True, nbins=4))
axB_inset.grid(axis="x", color="#D8D8D8", linewidth=0.55, alpha=0.55, zorder=0)
axB_inset.set_facecolor("white")

for spine in axB_inset.spines.values():
    spine.set_linewidth(0.8)
    spine.set_color("#777777")

# 添加计数值
offset = max(method_counts.max() * 0.03, 0.3)
for bar, value in zip(bars, method_counts.values):
    axB_inset.text(bar.get_width() + offset, bar.get_y() + bar.get_height()/2,
                   f"{int(value)}", ha="left", va="center",
                   fontsize=8.5, fontweight="bold", color=COL_TEXT)

axB_inset.set_xlim(0, method_counts.max() * 1.22 if method_counts.max() > 0 else 1)

add_panel_label(axB, "(b)", x=-0.07, y=1.055)

# ==========================================================
# (c) Tuyuksuyskiy Glacier 年际质量平衡
# ==========================================================
axC.plot(data_e["year"], data_e["oggm"], marker="s", linewidth=2.0,
         color=COL_OGGM, markerfacecolor="none", markeredgecolor=COL_OGGM,
         markeredgewidth=1.25, label="OGGM", markersize=7.5)

axC.plot(data_e["year"], data_e["wgms"], marker="^", linewidth=2.3,
         color=COL_WGMS, markerfacecolor="none", markeredgecolor=COL_WGMS,
         markeredgewidth=1.25, label="WGMS", markersize=7.5)

axC.axhline(0, color=COL_ZERO, linestyle="--", linewidth=1)
axC.set_ylabel("Mass balance (mm w.e.)", fontsize=15, fontweight="bold")
axC.legend(fontsize=12, loc="upper right", frameon=False, ncol=2,
           handlelength=2.2, columnspacing=1.2)
style_axes(axC, ticksize=12)
add_panel_label(axC, "(c)", x=-0.07, y=1.045)

# ==========================================================
# (d) Tuyuksuyskiy Glacier 相关性
# ==========================================================
corr_e = data_e["oggm"].corr(data_e["wgms"])
r2_e = corr_e ** 2

scatter_with_fit(
    axD,
    data_e["oggm"].values,
    data_e["wgms"].values,
    data_e["year"].values,
    xlabel="OGGM (mm w.e.)",
    ylabel="WGMS (mm w.e.)",
    box_text=f"Pearson $r$ = {corr_e:.3f}\n$R^2$ = {r2_e:.3f}",
    cmap="cividis"
)
add_panel_label(axD, "(d)", x=-0.12, y=1.045)

# ==========================================================
# (e) Urumqi Glacier No. 1 West Branch 年际质量平衡
# ==========================================================
axE.plot(data_w["year"], data_w["oggm"], marker="s", linewidth=2.0,
         color=COL_OGGM, markerfacecolor="none", markeredgecolor=COL_OGGM,
         markeredgewidth=1.25, label="OGGM", markersize=7.5)

axE.plot(data_w["year"], data_w["wgms"], marker="^", linewidth=2.3,
         color=COL_WGMS, markerfacecolor="none", markeredgecolor=COL_WGMS,
         markeredgewidth=1.25, label="WGMS", markersize=7.5)

axE.axhline(0, color=COL_ZERO, linestyle="--", linewidth=1)
axE.set_xlabel("Year", fontsize=15, fontweight="bold")
axE.set_ylabel("Mass balance (mm w.e.)", fontsize=15, fontweight="bold")
axE.legend(fontsize=12, loc="upper right", frameon=False, ncol=2,
           handlelength=2.2, columnspacing=1.2)
style_axes(axE, ticksize=12)
add_panel_label(axE, "(e)", x=-0.07, y=1.045)

# ==========================================================
# (f) Urumqi Glacier No. 1 West Branch 相关性
# ==========================================================
corr_w = data_w["oggm"].corr(data_w["wgms"])
r2_w = corr_w ** 2

scatter_with_fit(
    axF,
    data_w["oggm"].values,
    data_w["wgms"].values,
    data_w["year"].values,
    xlabel="OGGM (mm w.e.)",
    ylabel="WGMS (mm w.e.)",
    box_text=f"Pearson $r$ = {corr_w:.3f}\n$R^2$ = {r2_w:.3f}",
    cmap="cividis"
)
add_panel_label(axF, "(f)", x=-0.12, y=1.045)

# ==========================================================
# 统一调整
# ==========================================================
for ax in [axA, axB, axC, axD, axE, axF]:
    ax.set_axisbelow(True)

fig.align_ylabels([axA, axC, axE])
fig.align_ylabels([axB, axD, axF])

# ==========================================================
# 保存
# ==========================================================
out_path = r"D:/oggm/论文数据/wgms/Figure_Combined_Optimized.png"

plt.savefig(
    out_path,
    dpi=600,
    bbox_inches="tight",
    pad_inches=0.04,
    facecolor="white"
)
plt.show()

# ==========================================================
# 输出统计结果
# ==========================================================
print("Figure saved to:")
print(out_path)

print("\nGeodetic calibration statistics:")
print(f"R: {R:.3f}")
print(f"RMSE: {RMSE:.3f}")
print(f"Bias: {BIAS:.3f}")

print("\nTuyuksuyskiy Glacier:")
print(f"Pearson r: {corr_e:.3f}")
print(f"R-squared: {r2_e:.3f}")

print("\nUrumqi Glacier No. 1 West Branch:")
print(f"Pearson r: {corr_w:.3f}")
print(f"R-squared: {r2_w:.3f}")
```
